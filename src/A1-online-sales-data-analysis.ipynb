Uygulama Görevi 1: Online Satış Verisi Analizi
Veri Seti: satislar.csv dosyasını kullanarak aşağıdaki adımları takip ediniz.
1. Veri İçe Aktarma: pandas ile dosyayı df nesnesine oku.
2. Veri Analizi: head(), info() ve describe() ile yapıyı tanı.
3. Eksik Değer Yönetimi: Hangi sütunlarda eksik var? adet ve fiyat
sütunlarını medyan (aykırı değer direnci için) ile doldur.
4. Özellik Mühendisliği (Feature Engineering): toplam = adet * fiyat
sütununu üret.
5. Gruplama ve Analiz: groupby() ile kategori bazlı toplam satışları hesapla,
bar grafik çiz.
6. Zaman Serisi (Trend): tarih sütununu to_datetime ile dönüştür ve line
grafikle zaman analizini yap.
