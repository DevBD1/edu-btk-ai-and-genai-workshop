```python
import matplotlib .pyplot as plt
gunler = ["Pzt", "Sal", "Car"]
satis = [100 , 150, 90]
# Endustri standardi: fig, ax kullanimi
fig, ax = plt.subplots(figsize =(6, 4))
ax.plot(gunler, satis, marker='o', color='green', linestyle ='--')
ax.set_title("Gunluk Satis Trendi")
ax.set_xlabel("Gunler")
ax.set_ylabel("Satis (Adet)")
ax.grid(True)
plt.show()
```
