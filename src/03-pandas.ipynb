```python
veri = {
"cinsiyet": ["K", "E", "K", "E"],
"puan": [85, 70, 90, 60]
}
df = pd.DataFrame(veri)
grup = df.groupby("cinsiyet")["puan"].mean()
print(grup)
```

```python
istatistikler = df.groupby("cinsiyet").agg(
ortalama_puan = ("puan", "mean"),
en_yuksek_puan = ("puan", "max"),
kisi_sayisi = ("puan", "count")
)
print(istatistikler)
```
